Qn 1: Write a Python program to find the factorial of a number.

Prompt: Write a Python program to find the factorial of a given positive integer using a loop. Provide sample input and output.

```python
n = 5
factorial = 1

for i in range(1, n + 1):
    factorial *= i

print("Factorial:", factorial)
```

Logic: Multiply all integers from 1 to the given number to calculate its factorial.

Qn 2: Write a Python program to check whether a string is a palindrome.

Prompt: Write a Python program to check whether a given string is a palindrome. Ignore uppercase and lowercase differences and show sample input and output.

```python
text = "Madam"

if text.lower() == text.lower()[::-1]:
    print("Palindrome string")
else:
    print("Not a palindrome string")
```

Logic: Reverse the string and compare it with the original string after converting both to lowercase.

Qn 3: Write a Python program to remove duplicates from a list.

Prompt: Write a simple Python program to remove duplicate elements from a list while maintaining the original order. Provide sample input and output.

```python
numbers = [1, 2, 3, 2, 4, 1, 5]
unique = []

for number in numbers:
    if number not in unique:
        unique.append(number)

print("List after removing duplicates:", unique)
```

Logic: Check whether each element already exists in the new list. Add it only if it is not present.

Qn 4. Write a Python program to find the second largest number in a list

Prompt: Write a Python program to find the second largest distinct number in a list without using the sort() or sorted() functions. Use a simple approach and display the result.

```python
numbers = [10, 25, 8, 40, 30]

largest = second = float('-inf')

for num in numbers:
    if num > largest:
        second = largest
        largest = num
    elif num > second and num != largest:
        second = num

print("Second largest number:", second)
```

Logic: Maintain two variables to track the largest and second largest distinct values while traversing the list.

Qn 5. Write a Python program to check whether two strings are anagrams

Prompt: Create a Python program to check whether two given strings are anagrams of each other. Ignore spaces and uppercase/lowercase differences. Display an appropriate message.

```python
str1 = "Listen"
str2 = "Silent"

s1 = str1.replace(" ", "").lower()
s2 = str2.replace(" ", "").lower()

if sorted(s1) == sorted(s2):
    print("The strings are anagrams")
else:
    print("The strings are not anagrams")
```

Logic: Two strings are anagrams if they contain exactly the same characters with the same frequency

Qn 6. Write a Python program to calculate the power of a number without using the exponent operator

Prompt: Write a simple Python program to calculate the power of a number without using the ** operator or the pow() function. Use a loop and display the result.

```python
base = 3
exponent = 4

result = 1

for i in range(exponent):
    result *= base

print("Result:", result)
```

Logic: Multiply the base by itself repeatedly for the number of times specified by the exponent.

Qn 7. Write a Python program to count the occurrence of a particular character

Prompt: Write a Python program to count how many times a particular character appears in a given string. Ignore uppercase and lowercase differences.

```python
text = "Programming"
target = "m"

count = 0

for char in text.lower():
    if char == target.lower():
        count += 1

print("Character occurs:", count, "times")
```

Logic: Traverse the string and increment the counter whenever the character matches the target character.

Qn 8. Write a Python program to merge two lists alternately

Prompt: Write a Python program to merge two lists by taking one element from each list alternately. Assume both lists have the same number of elements.

```python
list1 = [1, 3, 5]
list2 = [2, 4, 6]

merged = []

for i in range(len(list1)):
    merged.append(list1[i])
    merged.append(list2[i])

print("Merged list:", merged)
```

Logic: Use a loop to take one element from each list and append them alternately to a new list.

Qn 9. Write a python program to find the common characters between two strings

Prompt: Write a Python program to find the unique characters that are present in both given strings. Ignore spaces and uppercase/lowercase differences.

```python
str1 = "Python"
str2 = "Phone"

common = []

for char in str1.lower():
    if char in str2.lower() and char not in common and char != " ":
        common.append(char)

print("Common characters:", common)
```

Logic: Check whether each character in the first string is present in the second string. Add matching characters only once.

Qn 10. Write a Python program to find the difference between the largest and smallest numbers

Prompt: Write a Python program to find the largest and smallest elements in a list and calculate their difference without using the max() and min() functions.

```python
numbers = [15, 8, 32, 5, 20]

largest = numbers[0]
smallest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

    if num < smallest:
        smallest = num

difference = largest - smallest

print("Largest:", largest)
print("Smallest:", smallest)
print("Difference:", difference)
```

Logic: Find the largest and smallest values by comparing each list element, then subtract the smallest from the largest.
