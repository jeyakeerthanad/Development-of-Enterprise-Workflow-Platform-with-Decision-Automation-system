Question 1: Write an SQL Query to count the number of sales in each city.

Prompt: Write an SQL query to count the number of sales records from each city. Display the city name and the corresponding number of sales.

```python
%%sql
SELECT city,
       COUNT(*) AS total_sales
FROM Sales
GROUP BY city;
```

Question 2: Write an SQL Query to find the total revenue generated by each product.

Prompt: Write an SQL query to calculate the total revenue generated by each product. Revenue should be calculated by multiplying quantity by price. Group the results by product name and arrange them in descending order of revenue.

```python
%%sql
SELECT product_name,
       SUM(quantity * price) AS total_revenue
FROM Sales
GROUP BY product_name
ORDER BY total_revenue DESC;
```

Question 3: Write an SQL Query to find the highest-priced product using a subquery.

Prompt: Write an SQL query using a subquery to find the product name and price of the product or products having the highest price in the Sales table.

```python
%%sql
SELECT DISTINCT product_name, price
FROM Sales
WHERE price = (
    SELECT MAX(price)
    FROM Sales
);
```

Question 4: Write an SQL Query to find products whose names start with the letter 'L'.

Prompt: Write an SQL query to display all products whose names start with the letter L using the LIKE operator.

```python
%%sql
SELECT *
FROM Sales
WHERE product_name LIKE 'L%';
```

Question 5: Write an SQL Query to find categories whose total revenue exceeds ₹10,000.

Prompt: Write an SQL query to calculate the total revenue generated by each product category. Display only those categories whose total revenue is greater than 10000. Use GROUP BY and HAVING clauses.

```python
%%sql
SELECT category,
       SUM(quantity * price) AS total_revenue
FROM Sales
GROUP BY category
HAVING SUM(quantity * price) > 10000;
```

Question 6: Write an SQL Query to display products belonging to selected categories.

Prompt: Write an SQL query to display products that belong to either the Fashion or Accessories category using the IN operator.

```python
%%sql
SELECT product_name, category, price
FROM Sales
WHERE category IN ('Fashion', 'Accessories');
```

Question 7: Write an SQL Query to calculate the total revenue for each city.

Prompt: Write an SQL query to calculate the total sales revenue generated in each city by multiplying quantity by price. Display the city and its total revenue.

```python
%%sql
SELECT city,
       SUM(quantity * price) AS total_revenue
FROM Sales
GROUP BY city;
```

Question 8: Write an SQL Query to display unique product categories.

Prompt: Write an SQL query to retrieve all distinct product categories from the Sales table without displaying duplicate category names.

```python
%%sql
SELECT DISTINCT category
FROM Sales;
```

Question 9: Write an SQL Query to display sales records in ascending order of price.

Prompt: Write an SQL query to display all sales records arranged in ascending order based on the product price.

```python
%%sql
SELECT *
FROM Sales
ORDER BY price ASC;
```

Question 10: Write an SQL Query to classify products based on their price.

Prompt: Write an SQL query using the CASE statement to classify products into three price categories: High Price for prices above 20000, Medium Price for prices between 2000 and 20000, and Low Price for prices below 2000.

```python
%%sql
SELECT product_name,
       price,
       CASE
           WHEN price > 20000 THEN 'High Price'
           WHEN price >= 2000 THEN 'Medium Price'
           ELSE 'Low Price'
       END AS price_category
FROM Sales;
```
